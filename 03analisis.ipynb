```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np

# Inicializamos el DataFrame a None.
df = None

print("--- 1. Carga de Datos y Limpieza de Columnas ---")
try:
    # Carga de datos
    # ¡Asegúrate de que 01database.csv esté subido al panel Archivos!
    df = pd.read_csv('01database.csv', sep=',', engine='python', on_bad_lines='skip')

    # PASO CRÍTICO: Limpieza de nombres de columnas
    df.columns = df.columns.str.strip().str.replace(' ', '_')

    if df is not None and not df.empty:
        print(f"✅ DataFrame cargado y columnas limpiadas. Filas: {len(df)}")
    else:
        print("❌ ERROR: El archivo está vacío o la carga falló silenciosamente.")
except Exception as e:
    print(f"❌ ERROR: La carga de datos falló: {e}")
    df = None
```
